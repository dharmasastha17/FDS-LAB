```python
# Import required libraries
import pandas as pd
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Create a DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Display first five rows
print("First 5 rows of the Iris dataset:")
print(df.head())

print("\nMeasures of Central Tendency\n")

# Mean
print("1. Mean:")
print(df.mean())

# Median
print("\n2. Median:")
print(df.median())

# Mode
print("\n3. Mode:")
print(df.mode().iloc[0])   # Display the first mode for each column

# Combined summary
summary = pd.DataFrame({
    'Mean': df.mean(),
    'Median': df.median(),
    'Mode': df.mode().iloc[0]
})

print("\nSummary of Central Tendency:")
print(summary)
```
