```python

# Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from scipy.spatial.distance import hamming, euclidean

df = pd.read_csv("/content/titanic ADS lab.csv")

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nSurvival Count:")
print(df["Survived"].value_counts())

print("\nPassenger Class Count:")
print(df["Pclass"].value_counts())

print("\nGender Count:")
print(df["Sex"].value_counts())

print("\nEmbarked Count:")
print(df["Embarked"].value_counts())

print("\nCorrelation Matrix (Numeric Columns):")
print(df.corr(numeric_only=True))

print("\nNominal Attributes:")
nominal = ["Sex", "Embarked", "Cabin", "Ticket", "Name"]
print(nominal)

print("\nNumeric Attributes:")
numeric = ["Age", "Fare", "SibSp", "Parch"]
print(numeric)

print("\nBinary Attributes:")
binary = ["Survived"]
print(binary)

temp = df[["Sex", "Embarked"]].dropna()

le1 = LabelEncoder()
le2 = LabelEncoder()

sex = le1.fit_transform(temp["Sex"])
embarked = le2.fit_transform(temp["Embarked"])

nominal_dissimilarity = hamming(sex, embarked)

print("\nNominal Dissimilarity (Sex vs Embarked):")
print(nominal_dissimilarity)

temp2 = df[["Age", "Fare"]].dropna()

age = temp2["Age"].values
fare = temp2["Fare"].values

numeric_dissimilarity = euclidean(age, fare)

print("\nNumeric Dissimilarity (Age vs Fare):")
print(numeric_dissimilarity)




```
