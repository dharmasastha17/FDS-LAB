```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
uploaded = files.upload()

df = pd.read_csv(list(uploaded.keys())[0])

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nSpecies Distribution:")
print(df["Species"].value_counts())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
sns.pairplot(
    df.drop(columns=["Id"]),
    hue="Species",
    diag_kind="hist"
)

plt.show()
```

```python
X = df.drop(columns=["Id", "Species"])
y = df["Species"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)
print("\nFeatures:")
print(X.columns.tolist())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Standardized Data:")
print(X_scaled[:5])
```

```python
pca_2 = PCA(n_components=2)

X_pca_2 = pca_2.fit_transform(X_scaled)

print("Shape after PCA with 2 components:", X_pca_2.shape)

print("\nExplained Variance Ratio:")
print(pca_2.explained_variance_ratio_)

print("\nTotal Explained Variance (%):")
print(pca_2.explained_variance_ratio_.sum() * 100)
```

```python
pca_2_df = pd.DataFrame(
    X_pca_2,
    columns=["PC1", "PC2"]
)

pca_2_df["Species"] = y.values

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_2_df,
    x="PC1",
    y="PC2",
    hue="Species",
    s=80
)

plt.title("PCA with 2 Principal Components")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```

```python
pca_3 = PCA(n_components=3)

X_pca_3 = pca_3.fit_transform(X_scaled)

print("Shape after PCA with 3 components:", X_pca_3.shape)

print("\nExplained Variance Ratio:")
print(pca_3.explained_variance_ratio_)

print("\nTotal Explained Variance (%):")
print(pca_3.explained_variance_ratio_.sum() * 100)
```

```python
from mpl_toolkits.mplot3d import Axes3D

pca_3_df = pd.DataFrame(
    X_pca_3,
    columns=["PC1", "PC2", "PC3"]
)

pca_3_df["Species"] = y.values

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

for species in pca_3_df["Species"].unique():
    subset = pca_3_df[pca_3_df["Species"] == species]

    ax.scatter(
        subset["PC1"],
        subset["PC2"],
        subset["PC3"],
        label=species,
        s=50
    )

ax.set_xlabel("Principal Component 1")
ax.set_ylabel("Principal Component 2")
ax.set_zlabel("Principal Component 3")
ax.set_title("PCA with 3 Principal Components")
ax.legend()

plt.show()
```

```python
variance_2 = pca_2.explained_variance_ratio_.sum() * 100
variance_3 = pca_3.explained_variance_ratio_.sum() * 100

comparison = pd.DataFrame({
    "Principal Components": [2, 3],
    "Explained Variance (%)": [variance_2, variance_3],
    "Information Loss (%)": [100 - variance_2, 100 - variance_3]
})

display(comparison)
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(
    ["2 Components", "3 Components"],
    [variance_2, variance_3],
    color=["skyblue", "orange"]
)

plt.ylabel("Explained Variance (%)")
plt.title("Explained Variance Comparison")
plt.ylim(0, 105)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 1,
        f"{height:.2f}%",
        ha="center"
    )

plt.show()
```

```python
pca_full = PCA()
pca_full.fit(X_scaled)

explained_variance = pca_full.explained_variance_ratio_ * 100
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(9, 5))

plt.bar(
    range(1, 5),
    explained_variance,
    alpha=0.7,
    label="Individual Explained Variance"
)

plt.plot(
    range(1, 5),
    cumulative_variance,
    marker="o",
    color="red",
    label="Cumulative Explained Variance"
)

plt.xlabel("Principal Components")
plt.ylabel("Explained Variance (%)")
plt.title("Scree Plot and Cumulative Explained Variance")
plt.xticks(range(1, 5))
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("PCA ANALYSIS RESULTS")
print("------------------------------")
print("Original Features:", X.shape[1])
print("Original Samples:", X.shape[0])

print("\nPCA with 2 Components")
print("Explained Variance: {:.2f}%".format(variance_2))
print("Information Loss: {:.2f}%".format(100 - variance_2))

print("\nPCA with 3 Components")
print("Explained Variance: {:.2f}%".format(variance_3))
print("Information Loss: {:.2f}%".format(100 - variance_3))

print("\nAdditional Variance Retained: {:.2f}%".format(
    variance_3 - variance_2
))
```
